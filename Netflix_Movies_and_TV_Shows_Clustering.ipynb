#### Chart - 6

````python
# Chart - 6 visualization code
# Clean and convert 'duration' for movies (in minutes) and TV shows (in seasons)

# For Movies: Extract numerical part and convert to int (minutes)
movies_data = data[data['type'] == 'Movie'].copy()
movies_data['duration_minutes'] = movies_data['duration'].str.extract('(\d+)').astype(int)

# For TV Shows: Extract numerical part and convert to int (seasons)
tvshows_data = data[data['type'] == 'TV Show'].copy()
tvshows_data['duration_seasons'] = tvshows_data['duration'].str.extract('(\d+)').astype(int)

# Calculate average duration for movies and TV shows
avg_movie_duration = movies_data['duration_minutes'].mean()
avg_tvshow_duration = tvshows_data['duration_seasons'].mean()

duration_comparison = pd.DataFrame({
    'Content Type': ['Movie (minutes)', 'TV Show (seasons)'] ,
    'Average Duration': [avg_movie_duration, avg_tvshow_duration]
})

# Plotting
plt.figure(figsize=(8, 6))
sns.barplot(x='Content Type', y='Average Duration', data=duration_comparison, palette='coolwarm')
plt.title('Average Duration of Movies vs. TV Shows on Netflix')
plt.ylabel('Average Duration')
plt.show()
````

##### 1. Why did you pick the specific chart?

A bar chart is suitable here because it allows for a direct comparison between two distinct categories (movies and TV shows) based on a single numerical metric (average duration). It clearly illustrates the difference in duration, even though the units are different (minutes vs. seasons), as the chart's purpose is to show the average length characteristic of each type.

##### 2. What is/are the insight(s) found from the chart?

From this chart, we can gain the following insights:

*   **Movies are generally longer than TV shows (in their respective units)**: The average movie duration (in minutes) is significantly higher than the average TV show duration (in seasons). While these are different units, it qualitatively indicates that individual movie productions are typically one-off longer content pieces, whereas TV shows are structured into multiple, shorter parts (seasons).
*   **Netflix's Content Strategy**: This difference reflects the inherent nature of the content types and Netflix's approach to offering both full-length cinematic experiences and episodic series. Movies provide immediate, complete narratives, while TV shows offer extended engagement over time.

## ***5. Hypothesis Testing***

### Based on your chart experiments, define three hypothetical statements from the dataset. In the next three questions, perform hypothesis testing to obtain final conclusion about the statements through your code and statistical testing.

#### 2. Perform an appropriate statistical test.

````python
from scipy.stats import chi2_contingency

# Convert 'date_added' to datetime objects, coercing errors to NaT
data['date_added'] = pd.to_datetime(data['date_added'], errors='coerce')

# Drop rows where 'date_added' could not be parsed (NaT values) or other NaNs introduced by previous wrangling
data.dropna(subset=['date_added'], inplace=True)

# Extract month_added and year_added for plotting (and for this hypothesis test)
data['month_added'] = data['date_added'].dt.month
data['year_added'] = data['date_added'].dt.year

# Define a cutoff year to divide the data into two periods
# Based on Chart 2, a shift is visible around 2017-2018. Let's use 2017.
cutoff_year = 2017

# Create a 'period' column based on 'year_added'
data['period'] = data['year_added'].apply(lambda x: 'Before 2017' if x < cutoff_year else '2017 Onwards')

# Create a contingency table
contingency_table = pd.crosstab(data['type'], data['period'])
print("Contingency Table (Content Type vs. Period):\n", contingency_table)

# Perform the Chi-squared test for independence
chi2, p_value, dof, expected = chi2_contingency(contingency_table)

print(f"\nChi-squared statistic: {chi2:.2f}")
print(f"P-value: {p_value:.3f}")

# Interpret the results
alpha = 0.05
if p_value < alpha:
    print("\nConclusion: Reject the null hypothesis. There is a statistically significant difference in the proportion of TV shows versus movies added to Netflix across the two time periods.")
    print("This supports the alternate hypothesis that Netflix has been increasingly focusing on TV shows relative to movies in more recent years.")
else:
    print("\nConclusion: Fail to reject the null hypothesis. There is no statistically significant difference in the proportion of TV shows versus movies added to Netflix across the two time periods.")
    print("This does not support the alternate hypothesis.")
````

##### Which statistical test have you done to obtain P-Value?

I have performed a **Chi-squared test for independence**.

##### Why did you choose the specific statistical test?

The Chi-squared test for independence is appropriate here because:

1.  **Nature of Variables**: We are examining the relationship between two categorical variables: `type` (Movie or TV Show) and `period` (Before 2017 or 2017 Onwards).
2.  **Hypothesis**: The goal is to determine if the observed distribution of content types across these two periods is significantly different from what would be expected if there were no relationship between the period and the type of content added.
3.  **Data Type**: The test works with frequency counts in a contingency table, which is exactly what we have by cross-tabulating `type` and `period`.

#### What all multi-label categorical encoding techniques have you used and why?

For multi-label categorical features such as `country`, `director`, `cast`, and `listed_in`, a common and effective approach for clustering and machine learning models is to use a combination of **splitting/expanding** and **one-hot encoding**.

**Technique:**
1.  **Splitting and Expanding**: For each of these columns, the entries are strings containing multiple values separated by commas (e.g., 'United States, India', 'Actor 1, Actor 2'). We first split these strings into individual labels.
2.  **Exploding**: After splitting, we use `DataFrame.explode()` to transform each element of a list-like entry into a separate row. This creates a longer DataFrame where each original entry with multiple labels is represented by multiple rows, one for each label.
3.  **Filtering Top Categories**: Due to the high cardinality (many unique values) in columns like `director`, `cast`, and `country`, directly one-hot encoding all unique values would result in an extremely wide and sparse DataFrame, which can be computationally expensive and may not be beneficial for model performance. Therefore, we will identify and select a reasonable number of the most frequent categories (e.g., top 50, top 100, or a threshold based on frequency) for each of these columns.
4.  **One-Hot Encoding**: After filtering, we will apply one-hot encoding to these selected top categories. This will create binary (0 or 1) columns for each chosen director, cast member, country, or genre, indicating their presence in the content's entry.

**Why these techniques?**
*   **Preserves Information**: Splitting and exploding ensures that all individual labels are captured and considered.
*   **Handles Multi-label Nature**: Standard one-hot encoding directly on comma-separated strings would treat the entire string as a single category, losing the individual components. This approach correctly represents the multi-label nature.
*   **Manages Dimensionality**: Filtering by top categories helps to control the dimensionality of the feature space, making the dataset more manageable for clustering algorithms while retaining the most impactful information.
*   **Model Compatibility**: One-hot encoded features are suitable for a wide range of machine learning algorithms, including clustering techniques that rely on distance metrics (e.g., K-Means).

We will apply this strategy to `country`, `director`, `cast`, and `listed_in` to prepare them for clustering.

````python
# Function to process multi-label columns and one-hot encode top N values
def process_multi_label(df, column_name, top_n=100):
    # Drop rows where the column is NaN before splitting
    df_temp = df.dropna(subset=[column_name]).copy()

    # Split string by comma and expand into new rows (explode)
    exploded_data = df_temp.assign(temp_col=df_temp[column_name].str.split(', ')).explode('temp_col')

    # Get value counts of individual labels
    value_counts = exploded_data['temp_col'].value_counts()

    # Select top N categories
    top_categories = value_counts.head(top_n).index.tolist()

    # Create one-hot encoded columns for the top categories
    for category in top_categories:
        # Ensure the column name is safe for DataFrame (replace spaces/special chars if needed)
        safe_category_name = f'{column_name}_{category.replace(" ", "_").replace("-", "_").replace("&", "and").replace("/", "_")}'
        df[safe_category_name] = df[column_name].apply(lambda x: 1 if isinstance(x, str) and category in x.split(', ') else 0)
    return df

# Apply to 'country' (top 20 for example, as per EDA insights)
data = process_multi_label(data, 'country', top_n=20)

# Apply to 'listed_in' (top 20 genres)
data = process_multi_label(data, 'listed_in', top_n=20)

# For 'director' and 'cast', considering their high cardinality, we might start with a smaller top_n
# or use a different strategy later if dimensionality becomes an issue. Let's use top 50 for now.
data = process_multi_label(data, 'director', top_n=50)
data = process_multi_label(data, 'cast', top_n=50)

# Drop original multi-label columns, as they've been featurized
data.drop(['country', 'listed_in', 'director', 'cast'], axis=1, inplace=True)

print("DataFrame after processing multi-label features:")
print(data.head())
print("\nUpdated DataFrame Info:")
data.info()
````

### 5. Data Transformation

#### Do you think that your data needs to be transformed? If yes, which transformation have you used. Explain Why?

Yes, our data often needs transformation before applying machine learning algorithms, especially clustering algorithms that rely on distance metrics (like K-Means). Here's why and what transformations we might consider:

**Reasons for Data Transformation:**
1.  **Scaling for Distance-Based Algorithms**: Features with larger numerical ranges can dominate the distance calculations in algorithms like K-Means. To prevent this, all numerical features should be scaled to a similar range.
2.  **Handling Skewness / Normality**: Some algorithms perform better when numerical features are normally distributed. If a feature is highly skewed, transformations like logarithmic, square root, or Box-Cox transformations can help make its distribution more symmetric.
3.  **Feature Interaction / Non-linearity**: For some models, creating polynomial features or other non-linear transformations can capture complex relationships that linear models might miss. However, for clustering, the primary focus is typically on scaling and handling skewness.

**Transformations We Will Use:**

1.  **Standardization (Z-score normalization)**: This is the most common scaling technique for algorithms sensitive to feature magnitudes. It transforms the data to have a mean of 0 and a standard deviation of 1. It's particularly useful when the features have different units or widely varying scales.
    *   **Why**: Our `feature_matrix` contains `release_year`, `year_added`, `month_added`, `movie_duration_min`, `tv_show_seasons`, and hundreds of binary one-hot encoded features, alongside the TF-IDF features (which are already normalized to some extent but still have varying scales). Standardizing the numerical features ensures that features like `release_year` (e.g., 2000s) don't overpower the binary features (0 or 1) or the TF-IDF scores (typically 0 to 1 with many near 0).

2.  **No explicit non-linear transformations** will be applied at this stage for numerical features like `release_year` or `duration` as they represent factual quantities, and skewness is less of a concern for clustering than it would be for regression, for example. The primary goal is to ensure comparable scales.

First, let's identify the columns that need scaling. Our TF-IDF features are already scaled internally by the `TfidfVectorizer` (they are usually between 0 and 1). ### 7. Dimesionality Reduction

#### Do you think that dimensionality reduction is needed? Explain Why?



Yes, dimensionality reduction is highly recommended and often necessary for our `feature_matrix` due to its high dimensionality (currently `4804 rows x 3037 columns`). Here's why:

1.  **Curse of Dimensionality**: In high-dimensional spaces, data points become sparse, making it difficult for distance-based algorithms (like K-Means) to effectively identify meaningful clusters. Distances between data points tend to converge, losing their discriminative power.
2.  **Computational Efficiency**: A large number of features significantly increases the computational cost and time required for clustering algorithms. Reducing dimensions can make the process much faster.
3.  **Noise Reduction**: Many of the dimensions, especially from TF-IDF, might represent noise or less important features. Dimensionality reduction techniques can help to extract the most significant underlying patterns and reduce the impact of noisy features.
4.  **Improved Clustering Performance**: By focusing on the most relevant features, clustering algorithms can often achieve better separation and more coherent clusters.
5.  **Visualization**: Reduced dimensions (e.g., to 2 or 3 components) allow for easier visualization of clusters, aiding in interpretation.

**Technique to be used: Principal Component Analysis (PCA)**

PCA is a widely used linear dimensionality reduction technique that transforms the data to a new set of orthogonal variables called principal components. These components capture the maximum variance in the data, with the first component capturing the most variance, the second the second most, and so on.

*   **Why PCA?**: PCA is suitable here because it's effective for numerical data (our `scaled_feature_matrix` is numerical) and can handle the high correlation that might exist among some of our one-hot encoded and TF-IDF features. It will help us retain the most significant information while drastically reducing the number of features, making the clustering task more manageable and effective.

````python
# Implement Dimensionality Reduction using PCA
from sklearn.decomposition import PCA

pca = PCA(n_components=None) # Start with None to get explained variance ratio for all components
pca.fit(scaled_feature_matrix)

# Plotting the explained variance ratio to find the elbow point
plt.figure(figsize=(10, 6))
plt.plot(np.cumsum(pca.explained_variance_ratio_))
plt.xlabel('Number of Components')
plt.ylabel('Cumulative Explained Variance')
plt.title('Explained Variance by Principal Components')
plt.grid(True)
plt.show()


n_components = np.where(np.cumsum(pca.explained_variance_ratio_) >= 0.90)[0][0] + 1
print(f"Number of components to explain 90% variance: {n_components}")

pca = PCA(n_components=n_components)
principal_components = pca.fit_transform(scaled_feature_matrix)

# Create a DataFrame for the principal components
pca_feature_matrix = pd.DataFrame(data = principal_components, columns = [f'principal_component_{i+1}' for i in range(n_components)])

print("Shape of PCA-transformed feature matrix:", pca_feature_matrix.shape)
print("Sample of PCA-transformed feature matrix (first 5 rows):\n", pca_feature_matrix.head())
````

#### Which dimensionality reduction technique have you used and why? (If dimensionality reduction done on dataset.)



We have used **Principal Component Analysis (PCA)** for dimensionality reduction.

**Why PCA?**

1.  **Handles High Dimensionality**: Our `scaled_feature_matrix` had over 3000 features, which is too high for efficient clustering and can lead to the 'curse of dimensionality'. PCA effectively reduces this to a smaller, more manageable set of features.
2.  **Captures Maximum Variance**: PCA identifies the principal components, which are orthogonal (uncorrelated) linear combinations of the original features. Each subsequent component captures the maximum remaining variance, ensuring that the most important information (patterns) in the data is retained.
3.  **Noise Reduction**: By discarding components that explain very little variance, PCA helps in filtering out noise and redundant information, which can improve the performance and interpretability of subsequent clustering algorithms.
4.  **Improved Computational Efficiency**: The reduced number of features (`n_components`) significantly speeds up the clustering process.
5.  **Suitability for Clustering**: For distance-based clustering algorithms like K-Means, it's crucial that all features contribute meaningfully and are not highly correlated. PCA helps decorrelate features and provides a more compact representation suitable for these algorithms.

We selected the number of components to explain **90%** of the total variance in the dataset, balancing information retention with dimensionality reduction.

### 8. Data Splitting

For clustering, which is an unsupervised learning task, the concept of splitting data into training and testing sets (as is common in supervised learning) is generally not applied in the same way.

**Why Data Splitting is Different in Unsupervised Learning:**

1.  **No Target Variable**: In supervised learning, data is split to train a model on labeled data and then evaluate its performance on unseen, labeled test data. Clustering, however, does not have a target variable or ground truth labels.
2.  **Discovering Intrinsic Structure**: The goal of clustering is to discover inherent patterns and structures within the entire dataset. 3.  **Model Evaluation**: Evaluation of clustering models relies on intrinsic metrics (like Silhouette, Calinski-Harabasz, Davies-Bouldin scores) that assess the quality of the clusters found within the data itself, rather than predictive performance on unseen data.

Therefore, for this project, the clustering algorithms (K-Means, Agglomerative, DBSCAN) were applied to the entire `pca_feature_matrix` (which represents our preprocessed and dimensionality-reduced dataset). The focus is on finding meaningful groupings across the complete available data.

In the context of unsupervised clustering, the concept of an "imbalanced dataset" doesn't directly apply in the same way it does for supervised classification problems.

````python
# ML Model - 3 Implementation: DBSCAN Clustering

from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.cluster import DBSCAN
from sklearn.metrics import silhouette_score, calinski_harabasz_score, davies_bouldin_score
import numpy as np
import pandas as pd

# --- Re-running Data Scaling and PCA to ensure pca_feature_matrix is defined ---
# This block is re-included as pca_feature_matrix was not found in the environment.

# Identify numerical columns that need scaling
columns_to_scale = [
    'release_year',
    'year_added',
    'month_added',
    'movie_duration_min',
    'tv_show_seasons'
]

# Initialize the StandardScaler
scaler = StandardScaler()

# Apply scaling to the identified numerical columns in a copy of the feature_matrix
# Assuming 'feature_matrix' is available from previous successful executions.
# If 'feature_matrix' is not defined, further upstream cells need to be re-executed.
scaled_feature_matrix = feature_matrix.copy()
scaled_feature_matrix[columns_to_scale] = scaler.fit_transform(scaled_feature_matrix[columns_to_scale])

# Initialize PCA to find optimal number of components
pca = PCA(n_components=None)
pca.fit(scaled_feature_matrix)

# Determine number of components for 90% variance
n_components = np.where(np.cumsum(pca.explained_variance_ratio_) >= 0.90)[0][0] + 1

# Apply PCA with the determined number of components
pca = PCA(n_components=n_components)
principal_components = pca.fit_transform(scaled_feature_matrix)

# Create a DataFrame for the principal components
pca_feature_matrix = pd.DataFrame(data = principal_components, columns = [f'principal_component_{i+1}' for i in range(n_components)])

# --- End of re-included Data Scaling and PCA ---

# Initialize DBSCAN model with chosen parameters
# 'eps' is the maximum distance between two samples for one to be considered as in the neighborhood of the other.
# 'min_samples' is the number of samples in a neighborhood for a point to be considered as a core point.
# These parameters often require tuning based on the dataset's characteristics.

dbscan_model = DBSCAN(eps=0.5, min_samples=10)

# Fit the Algorithm and Predict on the model
clusters_dbscan = dbscan_model.fit_predict(pca_feature_matrix)

# Add the cluster labels to our original data DataFrame (or a copy)
# Assuming 'data_with_clusters' is available from previous K-Means execution.
data_with_clusters['cluster_label_dbscan'] = clusters_dbscan

print("First 5 rows of data with DBSCAN Clustering labels:\n", data_with_clusters[['title', 'type_Movie', 'type_TV Show', 'cluster_label_dbscan']].head())

# DBSCAN can produce noise points (labeled -1). We should filter these for silhouette score.
# Calinski-Harabasz and Davies-Bouldin scores require at least 2 clusters and no noise points (or filtered out noise).

# Filter out noise points for evaluation metrics
valid_points_mask = (clusters_dbscan != -1)
if np.sum(valid_points_mask) > 1 and len(np.unique(clusters_dbscan[valid_points_mask])) > 1:
    silhouette_avg_dbscan = silhouette_score(pca_feature_matrix[valid_points_mask], clusters_dbscan[valid_points_mask])
    calinski_harabasz_dbscan = calinski_harabasz_score(pca_feature_matrix[valid_points_mask], clusters_dbscan[valid_points_mask])
    davies_bouldin_dbscan = davies_bouldin_score(pca_feature_matrix[valid_points_mask], clusters_dbscan[valid_points_mask])
else:
    silhouette_avg_dbscan = float('nan')
    calinski_harabasz_dbscan = float('nan')
    davies_bouldin_dbscan = float('nan')

print(f"\nNumber of clusters found (excluding noise): {len(np.unique(clusters_dbscan[valid_points_mask]))}")
print(f"Number of noise points: {np.sum(clusters_dbscan == -1)}")
print(f"\nSilhouette Score (DBSCAN): {silhouette_avg_dbscan:.3f}")
print(f"Calinski-Harabasz Score (DBSCAN): {calinski_harabasz_dbscan:.3f}")
print(f"Davies-Bouldin Score (DBSCAN): {davies_bouldin_dbscan:.3f}")
````

The most critical aspect of applying DBSCAN is the selection of its two core hyperparameters: `eps` and `min_samples`.

1.  **`min_samples`**: This can be estimated based on the dimensionality of the data or domain knowledge. A common rule of thumb is `min_samples >= dimensionality + 1`, or `min_samples = 2 * dimensionality`. For our PCA-transformed data, a higher `min_samples` might be needed.
2.  **`eps`**: This is often determined by plotting the k-distance graph (distance to the k-th nearest neighbor for each point, where k = `min_samples`) and identifying the 'elbow' point, which indicates a good separation between denser and sparser regions.

### 1. Which Evaluation metrics did you consider for a positive business impact and why?



For unsupervised clustering models like K-Means, Agglomerative Clustering, and DBSCAN, traditional evaluation metrics (like accuracy, precision, recall) are not applicable as there is no 'ground truth' or labeled data. Instead, we rely on **intrinsic evaluation metrics** that assess the quality of the clusters based on the data's internal structure. The metrics considered for a positive business impact are:

1.  **Silhouette Score**:
    *   **Description**: Measures how similar an object is to its own cluster (cohesion) compared to other clusters (separation). It ranges from -1 to 1. A score close to 1 indicates that the object is well-matched to its own cluster and poorly matched to neighboring clusters. A score around 0 means the object is on or very close to the decision boundary between two neighboring clusters. Negative values suggest objects might have been assigned to the wrong cluster.
    *   **Business Impact**: A higher Silhouette Score indicates more distinct and well-separated clusters. For Netflix, this translates to clearer content categories, which is highly beneficial for:
        *   **Targeted Recommendations**: If clusters are distinct, recommendations within a cluster are more likely to be relevant.
        *   **Content Strategy**: Clearly defined clusters can inform content acquisition and production teams about specific niches or broader categories that resonate with users.
        *   **User Experience**: Users can more easily navigate and discover content if categories are intuitive and not heavily overlapping.

2.  **Calinski-Harabasz Score (Variance Ratio Criterion)**:
    *   **Description**: Computes the ratio of between-cluster dispersion to within-cluster dispersion. A higher score generally relates to a model with better-defined clusters.
    *   **Business Impact**: Similar to the Silhouette Score, a higher Calinski-Harabasz score implies a better partition of the dataset. For business, it suggests that the identified content segments are internally cohesive and externally distinguishable, making them more actionable for strategic planning and personalized user experiences.

3.  **Davies-Bouldin Score**:
    *   **Description**: Measures the average similarity ratio between each cluster and its most similar cluster. Similarity is defined as the ratio of within-cluster distances to between-cluster distances. A lower Davies-Bouldin score indicates better clustering, meaning clusters are more separated and compact.
    *   **Business Impact**: A lower Davies-Bouldin score means clusters are more distinct from each other and internally more uniform. This is crucial for Netflix to create meaningful content groupings, avoid recommending irrelevant content due to ambiguous cluster boundaries, and optimize marketing efforts towards well-defined audience segments.

By considering these metrics, we aim to identify a clustering solution that provides clear, actionable insights into the Netflix content library, thereby enabling more effective content recommendations, targeted marketing, and strategic content acquisition decisions for positive business growth.

### 3. Explain the model which you have used and the feature importance using any model explainability tool?



**Chosen Model: Agglomerative Clustering**

Agglomerative Clustering is a **hierarchical clustering algorithm** that builds a hierarchy of clusters. It follows a 'bottom-up' approach:

1.  **Initialization**: Each data point is initially considered as a single, individual cluster.
2.  **Merging**: In each step, the two closest clusters are merged into a new, larger cluster.
3.  **Iteration**: This process continues iteratively until all data points belong to a single large cluster, or until a predefined stopping criterion (like reaching a specific number of clusters) is met.

The 'closeness' or 'distance' between clusters is determined by a **linkage criterion**. For our model, we used the **'ward' linkage**, which merges the pair of clusters that leads to the minimum increase in the total within-cluster variance. This linkage method is generally robust and works well with Euclidean distances, which are implicit after our data scaling and PCA.

**Feature Importance and Explainability for Clustering**

For unsupervised clustering models, there isn't a direct concept of 'feature importance' in the same way as in supervised models (e.g., feature weights in linear models or impurity reduction in tree-based models). Model explainability tools like SHAP or LIME are typically designed for supervised tasks to explain individual predictions. However, we can infer the importance of features by understanding how they contribute to the formation and characteristics of the clusters.

Our clustering was performed on the **PCA-transformed feature matrix**, which itself is a reduced representation of our original engineered features. These original features can be broadly categorized as:

1.  **Textual Features**: Derived from `description` and `title` via TF-IDF vectorization. These capture the semantic content of the titles.
2.  **Categorical Features (One-Hot Encoded)**: These include binary indicators for:
    *   `type` (Movie/TV Show)
    *   `rating` (e.g., TV-MA, PG-13)
    *   Top `country` of production
    *   Top `listed_in` genres
    *   Top `director`s
    *   Top `cast` members
3.  **Numerical Features (Scaled)**:
    *   `release_year`, `year_added`, `month_added`
    *   `movie_duration_min`, `tv_show_seasons`

**Inferring Feature Importance (Explaining Cluster Characteristics):**

To understand which original features are important to the Agglomerative Clusters, we would typically perform a **post-clustering analysis**:

*   **Analyze Cluster Profiles**: Examine the average (for numerical features) or most frequent (for categorical features) values within each cluster. For example:
    *   If one cluster predominantly contains content where `type_Movie` is 1, `listed_in_Dramas` is 1, and `country_United_States` is 1, with a high average `movie_duration_min`, then these features are highly important in defining that cluster.
    *   Another cluster might be characterized by `type_TV Show`, specific `director`s, and `listed_in_International_TV_Shows`.
*   **Principal Component Loadings**: Since PCA was used, the principal components are linear combinations of the original scaled features. By examining the 'loadings' (coefficients) of the original features onto the principal components that explain most of the variance, we can understand which original features contribute most to the dimensions along which our clusters are formed.

In essence, the 'importance' of a feature in clustering is determined by its ability to differentiate one cluster from another. Features that show significant variation in their distribution across different clusters are considered more important. For Agglomerative Clustering on this dataset, it is likely that a combination of **semantic similarity (from TF-IDF of description/title)**, **content type (Movie/TV Show)**, **genre**, and **production country** are key drivers in forming distinct content groups, as these are the most descriptive features we engineered into the `feature_matrix`.
